# Channels and Bands
Bollinger Bands, Keltner Channels, and Donchian Channels

```python
import numpy as np, pandas as pd
from quantjourney_ti import TechnicalIndicators

n=250
idx = pd.date_range('2024-01-01', periods=n, freq='D')
close = np.cumsum(np.random.normal(0,1,n)) + 100
high = close + np.abs(np.random.normal(0.5,0.2,n))
low  = close - np.abs(np.random.normal(0.5,0.2,n))
df = pd.DataFrame({'high':high,'low':low,'close':close}, index=idx)

ti = TechnicalIndicators()
bb = ti.BB(df['close'], 20, 2.0)
kc = ti.KELTNER(df[['high','low','close']], 20, 10, 2.0)
dc = ti.DONCHIAN(df[['high','low']], 20)
bb.tail(), kc.tail(), dc.tail()
```
