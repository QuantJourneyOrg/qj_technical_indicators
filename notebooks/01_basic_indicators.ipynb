# Basic Indicators Demo
This notebook demonstrates simple usage of the TechnicalIndicators API.

```python
import numpy as np
import pandas as pd
from quantjourney_ti import TechnicalIndicators

# Synthetic close series
idx = pd.date_range('2024-01-01', periods=200, freq='D')
close = pd.Series(np.cumsum(np.random.normal(0,1,200)) + 100, index=idx)

ti = TechnicalIndicators()
sma20 = ti.SMA(close, 20)
ema20 = ti.EMA(close, 20)
rsi14 = ti.RSI(close, 14)
macd = ti.MACD(close, 12, 26, 9)
sma20.tail(), ema20.tail(), rsi14.tail(), macd.tail()
```
