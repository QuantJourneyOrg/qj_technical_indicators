# Streaming Indicators Demo

```python
import numpy as np, pandas as pd
from quantjourney_ti import TechnicalIndicators
from quantjourney_ti._streaming import StreamingIndicators

ti = TechnicalIndicators()
si = StreamingIndicators(max_buffer_size=1000)

idx = pd.date_range('2024-01-01', periods=200, freq='D')
series = pd.Series(np.cumsum(np.random.normal(0,1,200)) + 100, index=idx)
for ts, val in series.items():
    si.update_tick('SYM', ts, close=float(val))

batch_sma = ti.SMA(series, 20).iloc[-1]
stream_sma = si.states['SYM'].last_values['sma_20']
batch_sma, stream_sma
```
